```python
import pandas as pd
import plotly.express as px
import zipfile


with zipfile.ZipFile('meteorite-landings.csv (1).zip', 'r') as zip_ref:
    zip_ref.extractall()

df = pd.read_csv('meteorite-landings.csv')


df = df.dropna(subset=['reclat', 'reclong', 'mass', 'year'])
df = df[(df['year'] >= 1800) & (df['year'] <= 2026)]


print(" КОСМИЧЕСКАЯ СТАТИСТИКА NASA:")
print(f"Всего крупных метеоритов исследовано: {len(df[df['mass'] > 5000])} шт.")
print("-" * 50)


print(" ТОП-5 САМЫХ ТЯЖЕЛЫХ МЕТЕОРИТОВ В БАЗЕ:")
top_5 = df.nlargest(5, 'mass')[['name', 'year', 'mass']]
for idx, row in top_5.iterrows():

    mass_tons = row['mass'] / 1000000
    print(f" Метеорит '{row['name']}' ({int(row['year'])} г.) — Вес: {mass_tons:.2f} тонн!")
print("-" * 50)
# ----------------------------------------


df_map = df[df['mass'] > 5000]


fig_glob = px.scatter_geo(
    df_map,
    lat='reclat',
    lon='reclong',
    size='mass',
    color='year',
    color_continuous_scale=px.colors.sequential.Plasma,
    hover_name='name',
    hover_data=['year', 'mass'],
    projection="orthographic",
    title="1. Хронология и места падения крупных метеоритов на Землю"
)

fig_glob.update_geos(
    showcountries=True, countrycolor="RebeccaPurple",
    showocean=True, oceancolor="MidnightBlue",
    showlakes=True, lakecolor="Blue",
)
fig_glob.show()


fig_pie = px.pie(
    df_map,
    names='fall',
    title='2. Как были обнаружены метеориты: упали на глазах (Fell) vs найдены на земле (Found)',
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig_pie.show()
```
